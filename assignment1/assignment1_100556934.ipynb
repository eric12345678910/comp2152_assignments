```python
#d.	Using a loop, calculate the total workout minutes for each friend. 
    # Add new key-value pairs in the dictionary for each friend. 
    # For the key, use something like “Alex_Total”, and for the value, use the calculated total minutes.

temp_dic = {} # other way of initializing a dictionary than used previously

for key, value in workout_stats.items():
    total_hours = 0
    if isinstance( value, tuple):
        
        # Debugging: verify value is an int to prevent int iteration error
        for v in value:
            if isinstance(v, int):
                total_hours += v
            else:
                total_hours += int(v)
        
    total_key = key + "_Total"

    temp_dic[total_key] = total_hours
    print(f"{total_key} : total_hours: {total_hours}")


workout_stats.update(temp_dic)
    
```

```python
#e.	Create a 2-dimensional (nested) list called workout_list by extracting all the workout minutes from the dictionary. 
workout_list = []

# Iterate through dictionary
for key, value in workout_stats.items():
    
    # Verify values are a tuple (filter _Total values)
    if(isinstance(value, tuple)):
        time_list = []

        # Iterate through individual's recorded times
        for val in value:
            time_list.append(val)
        
        workout_list.append(time_list)

    else:
        continue
            
```

```python
#i.	Include a section at the end of your program where you print:
#	The friend with the highest total workout minutes.
#	The friend with the lowest total workout minutes.

totals = {}
for k, v in workout_stats.items():
    
    # Reduce stats to _Total values
    total = "_Total"
    if(k[-6:] == total):
        # Update new dictionary with total values
        totals.update({k[:-6]: v})
    else:
        continue

# Select highest and lowest values from total dict
for k, v in totals.items():

    if k == min(totals):
        print("\nLowest Total Workout Minutes: \n" + min(totals) + ": " + str(v) + "  min(s)\n")

    if k == max(totals):
        print("Highest Total Workout Minutes: \n" + max(totals) + ": " + str(v) + "  min(s)")
    
    else: 
        continue
     
```
